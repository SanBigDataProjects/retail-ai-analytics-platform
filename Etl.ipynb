```python
orders.columns = orders.columns.str.lower().str.replace(" ", "_")
print("Cleaned columns:", orders.columns.tolist())
```

```python
orders.dropna(inplace=True)
```

```python
orders.drop(columns=["product_id"],axis=1)
```

```python
orders["orders_date"] = pd.to_datetime(orders["order_date"])
```

```python
orders['order_month'] = orders['orders_date'].dt.month
orders["order_year"]= orders["orders_date"].dt.year
orders['order_quarter'] = orders['orders_date'].dt.quarter
```

```python
orders['discount']= orders['list_price'] * orders ['discount_percent']/100
```

```python
orders
```

```python
orders['discount'] = orders['list_price'] * orders['discount_percent'] / 100
orders['sale_price'] = orders['list_price'] - orders['discount']
orders['profit'] = orders['sale_price'] - orders['cost_price']
orders['revenue'] = orders['sale_price'] * orders['quantity']
orders['profit_margin'] = (orders['profit'] / orders['sale_price'] * 100).round(2)

# Quick check
print("New columns added:")
print(orders[['order_date', 'order_month', 'order_year', 
              'order_quarter', 'sale_price', 'revenue', 
              'profit', 'profit_margin']].head())
print("\nShape:", orders.shape)
```
